```python
import IPython
import torch
from effortless_config import Config
import numpy as np
import yaml

from ddsp.model import DDSP
from preprocess import preprocess
from interpolation import get_interpolated_output, load_model_from_weights, get_model_with_interpolated_weights

class args(Config):
    CONFIG = "config.yaml"

with open(args.CONFIG, "r") as config_file:
    config = yaml.safe_load(config_file)

sr =config["preprocess"]["sampling_rate"]
```

```python
path_exported_model_violin = "export/runs_test_full_20260604_194722_vn/ddsp_vn_pretrained.ts"
path_exported_model_flute = "export/runs_test_full_20260604_194722_fl/ddsp_fl_pretrained.ts"

path_weights_violin = "runs/test_full/20260604_194722/vn/state.pth"
path_weights_flute = "runs/test_full/20260604_194722/fl/state.pth"
```

For interpolation, will use the averaged mean loudness and std loudness of these models:

```python
mean_loudness = 0
std_loudness = 0
min_loudness = 0
max_loudness = np.inf

for path in [path_weights_violin, path_weights_flute]:
    path_config_file = path.replace("state.pth", "config.yaml")

    config_train = yaml.safe_load(open(path_config_file, "r"))

    mean_loudness += config_train["data"]["mean_loudness"]
    std_loudness += config_train["data"]["std_loudness"]

mean_loudness /= 2
std_loudness /= 2

print(f"Mean loudness: {mean_loudness}, Std loudness: {std_loudness}")
```

# Use the model for triggering notes

```python
model_violin_exported = torch.jit.load(path_exported_model_violin)
model_flute_exported = torch.jit.load(path_exported_model_flute)
```

```python
def make_envelope(frames_per_note, n_notes, attack=5, decay=15):
    env = np.full(frames_per_note, mean_loudness, dtype=np.float32)   # sustain at mean
    env[:attack]  = np.linspace(min_loudness, max_loudness, attack)         # attack
    env[-decay:]  = np.linspace(mean_loudness, min_loudness, decay)         # decay
    return np.tile(env, n_notes)
```

```python
sr = config["preprocess"]["sampling_rate"]
block_size = config["preprocess"]["block_size"]
secs_per_note = 1

# define the scale to be played
notes_hz = [261.63, 293.66, 329.63, 349.23, 392.00, 440.00, 493.88, 523.25]
frames_per_note = int(secs_per_note * sr / block_size)
total_frames = frames_per_note * len(notes_hz)

# generate envelope for each note
loudness_vals = make_envelope(frames_per_note, len(notes_hz))
loudness = torch.tensor(loudness_vals).reshape(1, total_frames, 1)

# generate tensor of pitch values for each note
pitch_vals = np.repeat(notes_hz, secs_per_note * sr / block_size)
pitch = torch.tensor(pitch_vals, dtype=torch.float32).reshape(1, total_frames, 1)
```

Play a scale with the violin model:

```python
audio_violin_scale = model_violin_exported(pitch, loudness).squeeze().detach().cpu().numpy()
IPython.display.Audio(audio_violin_scale, rate=config["preprocess"]["sampling_rate"])
```

Play a scale with the flute model:

```python
audio_flute_scale = model_flute_exported(pitch, loudness).squeeze().detach().cpu().numpy()
IPython.display.Audio(audio_flute_scale, rate=sr)
```

# Timbre transfer with exported models

```python
input_path = "../data/URMP/Dataset/01_Jupiter_vn_vc/AuSep_1_vn_01_Jupiter.wav"
x, p, l = preprocess(input_path, **config["preprocess"])
pitch_tensor = torch.from_numpy(p).float().view(1, -1, 1)
loudness_tensor = torch.from_numpy(l).float().view(1, -1, 1)
loudness_norm = (loudness_tensor - mean_loudness) / std_loudness
```

```python
audio_violin_exported = model_violin_exported(pitch, loudness).reshape(-1).detach().cpu().numpy()
IPython.display.Audio(audio_violin_exported, rate=sr)
```

```python
audio_flute_exported = model_flute_exported(pitch, loudness).reshape(-1).detach().cpu().numpy()
IPython.display.Audio(audio_flute_exported, rate=sr)
```

# Interpolate between Outputs

```python
model_violin_loaded = load_model_from_weights(path_weights_violin, config)
model_flute_loaded = load_model_from_weights(path_weights_flute, config)
```

### Original instruments

**Violin**

```python
audio_violin_loaded = model_violin_loaded.forward(pitch_tensor, loudness_norm).reshape(-1).detach().cpu().numpy()
IPython.display.Audio(audio_violin_loaded, rate=sr)
```

**Flute**

```python
audio_flute_loaded = model_flute_loaded.forward(pitch_tensor, loudness_norm).reshape(-1).detach().cpu().numpy()
IPython.display.Audio(audio_flute_loaded, rate=sr)
```

### Going through interpolation

Violin

```python
with torch.no_grad():
    interpolated_audio_low = get_interpolated_output(        
        model_violin_loaded, 
        model_flute_loaded, 
        mean_loudness,
        std_loudness,
        pitch_tensor, 
        loudness_tensor, 
        alpha=0.0,
        reverb=False
    )
IPython.display.Audio(interpolated_audio_low.reshape(-1).detach().cpu().numpy(), rate=sr)
```

Mid point

```python
with torch.no_grad():
    interpolated_audio_middle = get_interpolated_output(  
        model_violin_loaded, 
        model_flute_loaded, 
        mean_loudness,
        std_loudness,
        pitch_tensor, 
        loudness_tensor, 
        alpha=0.5
    )
IPython.display.Audio(interpolated_audio_middle.reshape(-1).detach().cpu().numpy(), rate=sr)
```

Flute

```python
interpolated_audio_high = get_interpolated_output(        
        model_violin_loaded, 
        model_flute_loaded, 
        mean_loudness,
        std_loudness,
        pitch_tensor, 
        loudness_tensor, 
        alpha=1.0,
        reverb=False
)
IPython.display.Audio(interpolated_audio_high.reshape(-1).detach().cpu().numpy(), rate=sr)
```
